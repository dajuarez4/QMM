# Figuras del paper: original y reproducción lado a lado

Se muestran **únicamente los 8 bloques de figuras activos del LaTeX que adjuntaste**, en su mismo orden. Los bloques 7 y 8 conservan sus tres imágenes apiladas. Las figuras comentadas y las gráficas adicionales del notebook de auditoría quedan fuera de esta selección.

**Izquierda:** imagen original de la carpeta principal del paper. **Derecha:** curvas calculadas con tu código QMM a **Λ = 300 MeV**, con los colores RGB originales, líneas continuas para Dieterici/Clausius, referencia vdW negra discontinua, límite conformal gris discontinuo y fuente Times New Roman cuando está disponible.

Este notebook vuelve a dibujar las tablas ya validadas; no recalcula ni cambia el EOS, el equilibrio beta ni los parámetros. **Run All** genera las comparaciones y las exporta a PNG/PDF.

Para mostrar el mismo tramo estelar que la figura de densidad central del paper se usa `nB,c / n0 ≤ 8`; el corte se puede cambiar abajo y no modifica las tablas completas. Las bandas observacionales permanecen visibles en el original; no se reconstruyen en la reproducción porque sus datos de contorno no fueron proporcionados.

```python
from pathlib import Path
import sys
from IPython.display import display
candidates=[q for p in [Path.cwd(),*Path.cwd().parents] for q in (p,p/'good_repo/QMM')]
ROOT=next(p.resolve() for p in candidates if (p/'scripts/paper_figures_side_by_side.py').exists())
sys.path.insert(0,str(ROOT/'scripts'))
import paper_figures_side_by_side as view
SOURCE=ROOT/'docs/paper_figures_source_snapshot.tex'  # Copia del LaTeX adjunto.
PAPER=ROOT.parents[1]/'Effects_of_Nuclear_Incompressibility_on_Quarkyonic_Neutron_Stars'
RESULTS=ROOT/'results/generated/paper_replication_lambda300'
OUT=ROOT/'results/generated/paper_figures_side_by_side'
MAX_CENTRAL_N_OVER_N0=8.0  # None permite mostrar toda la secuencia calculada.
blocks,reproductions=view.build(source=SOURCE,paper=PAPER,results=RESULTS,
                               output=OUT,central_max=MAX_CENTRAL_N_OVER_N0)
print(f'{len(blocks)} figuras activas; originales a la izquierda y reproducciones a la derecha.')
print('Exportaciones:',OUT)
```
